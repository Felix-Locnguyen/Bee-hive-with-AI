```python
from collections import Counter 
from pathlib import Path 
import pandas as pd 
import soundfile as sf # đọc tệp âm thanh (dạng mp3)
import os
```

```python
raw_path = "/home/loc-nguyen/PROJECT/AI/hive-management-ai/ai_engine/analyze_bee_sound/Data/AUDIO-BEE-DATA"
DATA_DIR = Path(raw_path.strip())

if not DATA_DIR.exists():
    print("Không tồn tại path:", DATA_DIR)
else:
    # Dùng rglob("*") nếu file nằm trong thư mục con (ví dụ theo nhãn/ngày), hoặc iterdir() nếu nằm trực tiếp
    files = [p.name for p in DATA_DIR.rglob("*") if p.is_file()]
    print(f"Tổng số file tìm thấy: {len(files)}")
    for name in files[:5]:
        print(name)
```

```python
files = sorted(os.listdir(DATA_DIR))
print("Số file:", len(files))
print("10 file đầu:", files[:10])

# Đọc thử 1 file .lab
for f in files:
  if f.endswith(".lab"):
    print("\n--- File .lab:", f)
    with open(os.path.join(DATA_DIR, f), encoding="utf-8", errors="ignore") as fp:
      print(fp.read())
    break
```
